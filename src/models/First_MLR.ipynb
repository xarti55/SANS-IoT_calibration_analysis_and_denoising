```python
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
# Reading the dataset
df = pd.read_csv(
    "../../data/data_Hw3_new.csv",
    sep=";",
    parse_dates=["Date"],
    date_format=lambda x: pd.to_datetime(x, format="%m/%d/%y %H:%M")
)
df .head()
```

```python
# Separate the information from Captor from the true values
# Captor data
X = df .iloc[:, 1:-1]
# True values
y = df .iloc[:, -1]
```

```python
# Number of samples to use for training, first 4 weeks (24*7*4)
n_train = 672

X_train = X.iloc[:n_train]
X_test  = X.iloc[n_train:]

y_train = y.iloc[:n_train]
y_test  = y.iloc[n_train:]
```

```python
# Create the MLR model 
reg = LinearRegression().fit(X_train, y_train)
```

```python
y_predicted = reg.predict(X_test)
```

```python
r2 = r2_score(y_test, y_predicted)
r2
```

```python
mae = mean_absolute_error(y_test, y_predicted)
mae
```

```python
mse = mean_squared_error(y_test, y_predicted)
rmse = float(np.sqrt(mse))
rmse
```
