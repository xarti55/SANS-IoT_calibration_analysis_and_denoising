```python
import numpy as np
import pandas as pd
from sklearn import datasets, ensemble
from sklearn.inspection import permutation_importance
from sklearn.utils.fixes import parse_version
from sklearn.model_selection import GridSearchCV
from sklearn.neighbors import KNeighborsRegressor
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor
from sklearn.ensemble import GradientBoostingRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
%matplotlib inline
import matplotlib.pyplot as plt
```

```python
df = pd.read_excel("../../data/data_Hw3_new.xlsx")
# Ensure timestamp column is datetime
df['timestamp'] = pd.to_datetime(df['timestamp'])
feature_cols = ['A0', 'A1', 'A2', 'A3', 'temperature', 'humidity']
target_col = 'O3'
```

```python
df['date'] = df['timestamp'].dt.date
df['hour'] = df['timestamp'].dt.hour
df = df.sort_values(by=["date", "hour"]).reset_index(drop=True)
```

```python
training_days = 28
training_hours = 28*training_days
X_train = df[feature_cols][:training_hours]
y_train = df[target_col][:training_hours]
X_test = df[feature_cols][training_hours:]
y_test = df[target_col][training_hours:]
```

```python
mlr_best_model = LinearRegression().fit(X_train, y_train)
print("-----MLR-----")
print("Training R2:", "{:.3f}".format(r2_score(mlr_best_model.predict(X_train), y_train)))
```

```python
y_predicted = mlr_best_model.predict(X_test)
print("-----MLR-----")
print("Testing R2:", "{:.3f}".format(r2_score(y_test, y_predicted)))
print("MAE:", "{:.3f}".format(mean_absolute_error(y_test, y_predicted)))
print("RMSE:", "{:.3f}".format(float(np.sqrt(mean_squared_error(y_test, y_predicted)))))
```

```python
param_grid = {'n_neighbors': list(range(1, 20))}
knr = KNeighborsRegressor(n_jobs=-1)

knn_grid = GridSearchCV(knr, param_grid, cv=5, scoring='r2', n_jobs=-1, return_train_score=True)
knn_grid.fit(X_train, y_train)

print("-----KNN-----")
print("Training R2:", "{:.3f}".format(knn_grid.best_score_))

knn_best_model = knn_grid.best_estimator_
print("Optimal N neigbors:", knn_grid.best_params_["n_neighbors"])
```

```python
y_predicted = knn_best_model.predict(X_test)
print("-----KNN-----")
print("Testing R2:", "{:.3f}".format(r2_score(y_test, y_predicted)))
print("MAE:", "{:.3f}".format(mean_absolute_error(y_test, y_predicted)))
print("RMSE:", "{:.3f}".format(float(np.sqrt(mean_squared_error(y_test, y_predicted)))))
```

```python
param_grid = {'max_depth': list(range(1, 25))}
rfr = RandomForestRegressor(n_jobs=-1, random_state=0)

rf_grid = GridSearchCV(rfr, param_grid, cv=5, scoring='r2', n_jobs=-1, return_train_score=True)
rf_grid.fit(X_train, y_train)

print("-----RF-----")
print("Training R2:", "{:.3f}".format(rf_grid.best_score_))

rf_best_model = rf_grid.best_estimator_

print("Optimal depth:", rf_grid.best_params_["max_depth"])
```

```python
y_predicted = rf_best_model.predict(X_test)
print("-----RF-----")
print("Testing R2:", "{:.3f}".format(r2_score(y_test, y_predicted)))
print("MAE:", "{:.3f}".format(mean_absolute_error(y_test, y_predicted)))
print("RMSE:", "{:.3f}".format(float(np.sqrt(mean_squared_error(y_test, y_predicted)))))
```

```python
param_grid = {'max_depth': list(range(1, 15))}
gbr = ensemble.GradientBoostingRegressor(n_estimators = 500, min_samples_split = 5, learning_rate = 0.01, random_state=0)

gb_grid = GridSearchCV(gbr, param_grid, cv=5, scoring='r2', n_jobs=-1, return_train_score=True)
gb_grid.fit(X_train, y_train)

print("-----GB-----")
print("Training R2:", "{:.3f}".format(gb_grid.best_score_))

gb_best_model = gb_grid.best_estimator_

print("Optimal depth:", gb_grid.best_params_["max_depth"])
```

```python
y_predicted = gb_best_model.predict(X_test)
print("-----RF-----")
print("Testing R2:", "{:.3f}".format(r2_score(y_test, y_predicted)))
print("MAE:", "{:.3f}".format(mean_absolute_error(y_test, y_predicted)))
print("RMSE:", "{:.3f}".format(float(np.sqrt(mean_squared_error(y_test, y_predicted)))))
```
